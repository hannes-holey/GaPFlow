Bayada-Chupin Cavitation Examples

```
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from GaPFlow.problem import Problem
from utils.plotting import plot_overview_1d, plot_overview_2d
from utils.topography import regen_twin_parabolic_slider, regen_twin_parabolic_slider_id

DATA_DIR = Path('../../tests/ref_data')
```

---
## Parabolic slider — Bayada-Chupin EoS

Single symmetric parabolic slider (hmin=25.4 µm, hmax=50.8 µm).  
Dirichlet ρ=ρ_l at both ends, periodic in y (effectively 1D).

**Reference:** Bayada et al. (2013).

<img src="./figures/Geometry_ParabolicSlider.png"
     alt="Parabolic slider geometry"
     width="250">

$\rightarrow$ **[YAML File](./examples/parabolic_slider_bayada.yaml)**

```
problem_par = Problem.from_yaml('./examples/parabolic_slider_bayada.yaml')
problem_par.run()
```

```
plot_overview_1d(
    problem_par,
    title='Parabolic slider — Bayada-Chupin',
    ref_csv=DATA_DIR / 'ref_parabolic_slider.csv',
    ref_label='Bayada',
    ref_x_scale=problem_par.grid['Lx'],
    ref_p_scale=1.0,
    p_unit='bar', p_divisor=1e5,
)
```

## Twin parabolic slider non-identical — Bayada-Chupin EoS

Two asymmetric parabolic sliders (hmin₁=30 µm, hmin₂=35 µm, hmax=60 µm)  
separated by a flat region and followed by a flat outlet.  
Dirichlet ρ=ρ_l at both ends, periodic in y.

**Reference:** Bayada et al. (Fig. 7).

<img src="./figures/Geometry_TwinParabolicSlider.png"
     alt="Twin parabolic slider geometry"
     width="250">

$\rightarrow$ **[YAML File](./examples/twin_parabolic_slider_bayada.yaml)**

```
problem_twin = Problem.from_yaml('./examples/twin_parabolic_slider_bayada.yaml')
regen_twin_parabolic_slider(problem_twin)
problem_twin.run()
```

```
plot_overview_1d(
    problem_twin,
    title='Twin parabolic slider non-identical — Bayada-Chupin',
    ref_csv=DATA_DIR / 'ref_twin_parabolic_slider_ori.csv',
    ref_label='Bayada',
    ref_x_scale=0.09,
    ref_p_scale=1e5,
    p_unit='bar', p_divisor=1e5,
)
```

## Twin parabolic slider identical — Bayada-Chupin EoS

Two identical symmetric parabolic sliders (hmax=50.2 µm, hmin=25.4 µm).  
High inlet pressure p_in=3.36e5 Pa, open outlet, periodic in y.

**Reference:** Sahlin et al. / Giacopini et al., J. Tribology 132 (2010), 041702.

<img src="./figures/Geometry_TwinParabolicSliderIdentical.png"
     alt="Twin identical parabolic slider geometry"
     width="250">

$\rightarrow$ **[YAML File](./examples/twin_parabolic_slider_id_smooth_bayada.yaml)**

```
problem_twin_id = Problem.from_yaml('./examples/twin_parabolic_slider_id_smooth_bayada.yaml')
regen_twin_parabolic_slider_id(problem_twin_id)
problem_twin_id.run()
```

```
plot_overview_1d(
    problem_twin_id,
    title='Twin parabolic slider identical — Bayada-Chupin',
    ref_csv=DATA_DIR / 'ref_twin_parabolic_slider_id.csv',
    ref_label='Sahlin',
    ref_x_scale=problem_twin_id.grid['Lx'],
    ref_p_scale=1e6,
    p_unit='MPa', p_divisor=1e6,
)
```
