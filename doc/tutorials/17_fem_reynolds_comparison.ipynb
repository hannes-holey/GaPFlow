# FEM NS vs FEM Reynolds Solver Comparison

Comparing the full Taylor-Hood P2P1 momentum solver (FEM NS) against the reduced,
no-squeeze scalar Reynolds equation solver (FEM Reynolds, `fem_solver.equations.reynolds: true`,
DOF `p` only, no `jx`/`jy`) on the same physical problems:
- **FEM NS**: Full FEM 2D momentum solver (jx, jy, p)
- **FEM Reynolds**: FEM 2D scalar Reynolds solver (p only)

```
from utils import (load_template, run_reynolds_comparison, plot_reynolds_comparison_rho,
                   print_reynolds_timing, run_reynolds_comparison_2d, plot_reynolds_comparison_rho_2d)
```

## Inclined Slider (Air, PL EOS)

Linear converging gap with Dirichlet density BC.

<img src="./figures/Geometry_InclinedSlider.png"
     alt="Inclined slider geometry"
     width="250">

$\rightarrow$ **[YAML File](../../tests/configs/inclined_slider.yaml)**

```
template = load_template('inclined_slider')
results_inclined = run_reynolds_comparison(template)
```

```
plot_reynolds_comparison_rho(results_inclined, 'Inclined Slider')
print_reynolds_timing(results_inclined)
```
