# Comparison of Explicit and Implicit Solvers

Here, we compare the predictions of the explicit and implicit solvers for different hydrodynamic test cases and assess their consistency.

Note that the referenced `yaml` files hold the configuration for the implicit solver. We use overrides for running the explicit solver `explicit_overrides` to adapt e.g. timestep and writing frequency.

```
from utils import load_template, run_all_solvers, plot_solver_comparison_rho_jx
```

## Inclined Slider

Inclined slider with Dirichlet density BC.

<img src="./figures/Geometry_InclinedSlider.png"
     alt="Inclined slider geometry"
     width="250">

$\rightarrow$ **[YAML File](../../tests/configs/inclined_slider.yaml)**

```
template = load_template('inclined_slider')
explicit_overrides={'numerics.adaptive': True, 'numerics.CFL': 0.2, 'options.write_freq': 1000}
results_inclined = run_all_solvers(template, explicit_overrides)
plot_solver_comparison_rho_jx(results_inclined)
```

## Journal Bearing

Cosine gap profile with fully periodic BC.

<img src="./figures/Geometry_JournalBearing.png"
     alt="Journal Bearing geometry"
     width="250">

$\rightarrow$ **[YAML File](../../tests/configs/journal_bearing.yaml)**

```
template = load_template('journal_bearing')
explicit_overrides = {'numerics.adaptive': True, 'numerics.CFL': 0.1, 'options.write_freq': 1000}
results_journal_bearing = run_all_solvers(template, explicit_overrides)
plot_solver_comparison_rho_jx(results_journal_bearing)
```
