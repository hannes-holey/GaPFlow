# FEM NS vs FEM Reynolds Solver Comparison — Cavitation

Comparing the full Taylor-Hood P2P1 momentum solver (FEM NS) against the reduced,
no-squeeze scalar Reynolds equation solver (FEM Reynolds), both with
Fischer-Burmeister cavitation enabled (`fem_solver.equations.cavitation: true`):
- **FEM NS**: Full FEM 2D momentum solver (jx, jy, p, theta)
- **FEM Reynolds**: FEM 2D scalar Reynolds solver (p, theta only)

```
from pathlib import Path

from utils import (load_template, run_reynolds_cav_comparison,
                   plot_reynolds_cav_comparison, print_reynolds_timing,
                   regen_twin_parabolic_slider_id)

DATA_DIR = Path('../../tests/ref_data')
```

## Parabolic Slider (Oil, DH EOS, Fischer-Burmeister Cavitation)

Parabolic gap profile with Dirichlet pressure BC, periodic in y (effectively 1D).

<img src="./figures/Geometry_ParabolicSlider13.png"
     alt="Parabolic slider geometry"
     width="250">

$\rightarrow$ **[YAML File](../../tests/configs/parabolic_slider_cav_rey.yaml)**

```
template = load_template('parabolic_slider_cav_rey')
results_parabolic_cav = run_reynolds_cav_comparison(template)
```

```
plot_reynolds_cav_comparison(results_parabolic_cav, 'Parabolic Slider (Cavitation)')
print_reynolds_timing(results_parabolic_cav)
```

## Twin Parabolic Slider, Identical (Oil, DH EOS, Fischer-Burmeister Cavitation)

Two identical symmetric parabolic pads (hmax=50.2 um, hmin=25.4 um), height profile
injected via regen_twin_parabolic_slider_id (not expressible via geometry.type alone).
High inlet pressure, periodic in y (effectively 1D).

<img src="./figures/Geometry_TwinParabolicSliderIdentical.png"
     alt="Twin parabolic slider (identical) geometry"
     width="250">

$\rightarrow$ **[YAML File](../../tests/configs/twin_parabolic_slider_id_cav_rey.yaml)**

```
template = load_template('twin_parabolic_slider_id_cav_rey')
results_twin_parabolic_id_cav = run_reynolds_cav_comparison(
    template, topo_fn=regen_twin_parabolic_slider_id)
```

```
plot_reynolds_cav_comparison(
    results_twin_parabolic_id_cav, 'Twin Parabolic Slider, Identical (Cavitation)',
    p_unit='MPa', p_divisor=1e6,
    ref_csv=DATA_DIR / 'ref_twin_parabolic_slider_id.csv', ref_label='Sahlin',
    ref_x_scale=1.0, ref_p_scale=1e6)
print_reynolds_timing(results_twin_parabolic_id_cav)
```
